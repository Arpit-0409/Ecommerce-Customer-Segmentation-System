```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
```

```python
trf = ColumnTransformer([
    ("ohe",OneHotEncoder(sparse_output=False),[0,1]),
    ("scale",StandardScaler(),slice(2,15))
] )
```

```python
X_transformed = trf.fit_transform(cart_data)
```

```python
X_transformed
```

```python
#VISUALIZING THE DATA
```

```python
from sklearn.decomposition import PCA
import numpy as np
```

```python
X_transformed.shape
```

```python
pca = PCA(n_components = 3)
```

```python




X_pca = pca.fit_transform(X_transformed)
```

```python
(pca.explained_variance_ratio_)*100
```

```python
#explained variance_ratio  - CUMULATIVE SUM OF PRINCIPAL COMPONENTS
np.cumsum(pca.explained_variance_ratio_)*100
```

```python
import plotly.express as px

px.scatter_3d(x = X_pca[:,0],y =X_pca[:,1],z =X_pca[:,2])

```

```python
#clustering algorthim - to form cluster 
```

```python
#Finding K VALUE 
#1. ELBOW METHOD
from sklearn.cluster import KMeans
wcss = []
for  k in range(1,11):
    kmeans = KMeans(n_clusters = k ,random_state = 42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)

```

```python
plt.plot(range(1,11),wcss,marker ="o")
plt.xlabel("K")
plt.ylabel("WCSS")
```

```python
#silhouette_score
```

```python
from sklearn.metrics import silhouette_score
```

```python
scores =[]
for k in range(2,11):
    kmeans = KMeans(n_clusters=k,random_state = 42)
    labels =kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca,labels)
    scores.append(score)
    
```

```python

plt.plot(range(2,11),scores,marker ="o")

plt.xlabel("K")
plt.ylabel("Silhuate_score")
```

```python
#combined plot
k_range = range(2,11)
fix , ax1 = plt.subplots(figsize=(5,5))
ax1.plot(k_range,wcss[:len(k_range)], marker="o",color="red")
ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()
ax2.plot(k_range,scores[:len(k_range)], marker="o",color="green")

ax2.set_ylabel("score")

```
